## Data visualization (Perform EDA)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/AI Job Market Dataset.csv')
df.head()
```

```python
df.shape
```

```python
df.count()
```

# 1. Which job in highst demand

```python
df.columns
```

```python
job_count = df.groupby('job_title')['job_openings'].sum().sort_values(ascending=False).reset_index()
job_count.rename(columns={'job_openings': 'Count'}, inplace=True)

plt.figure(figsize=(10,6))

plt.barh(
    job_count['job_title'],
    job_count['Count']
)

plt.title('Highest Demand Jobs')
plt.xlabel('Number of Openings')
plt.ylabel('Job Title')

# Highest demand at top
plt.gca().invert_yaxis()

plt.show()
```

```python
df.head()
```

#2. Which countries have the highest job demand?

```python
df.head()
```

```python
country_demand = df.groupby('country')['job_openings'].sum().sort_values(ascending=False).reset_index()
country_demand.rename(columns={'job_openings': 'job_count'}, inplace=True)

plt.figure(figsize=(10,5))

plt.bar(
    country_demand['country'].head(10),
    country_demand['job_count'].head(10)
)

plt.title("Top 10 Countries by Job Demand")
plt.xlabel("Country")
plt.ylabel("Number of Jobs")

plt.xticks(rotation=45)

plt.show()
```

# 3. Which industries are hiring the most demand

```python
hiring_demand = df.groupby('company_industry')['job_openings'].sum().sort_values(ascending=False).reset_index()

plt.figure(figsize=(10,5))

plt.bar(
    hiring_demand['company_industry'].head(10),
    hiring_demand['job_openings'].head(10)
)

plt.title("Top 10 Industries by Job Demand")
plt.xlabel("company_industry")
plt.ylabel("Number of job openings")

plt.xticks(rotation=45)

plt.show()
```

# 4. How does salary change with experience

```python
# Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df['years_experience'], df['salary'])

plt.title("Salary vs Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary (LPA)")
plt.grid(True)

plt.show()
```

```python
sns.lineplot(x='years_experience', y ='salary', data=df)
plt.title('experiece wise salary')
plt.show()
```

# 5. Which technical skill are most requested

```python
from collections import Counter

# Identify skill columns
skill_columns = ['skills_python', 'skills_sql', 'skills_ml', 'skills_deep_learning', 'skills_cloud']

# Initialize a dictionary to store skill counts
skill_counts_dict = {}

# Calculate counts for each skill
for col in skill_columns:
    # The sum of a boolean-like (0/1) column directly gives the count of '1's
    # We extract the skill name by removing the 'skills_' prefix
    skill_counts_dict[col.replace('skills_', '')] = df[col].sum()

# Convert to DataFrame
skill_df = pd.DataFrame(
    skill_counts_dict.items(),
    columns=['Skill', 'Count']
)

# Sort descending
skill_df = skill_df.sort_values(
    by='Count',
    ascending=False
)

print(skill_df)
```

```python
df.head()
```

```python
plt.figure(figsize=(10,6))

plt.barh(
    skill_df['Skill'],
    skill_df['Count']
)

plt.title('Most Requested Technical Skills')
plt.xlabel('Number of Job Postings')
plt.ylabel('Skills')

plt.gca().invert_yaxis()

plt.show()
```

# 6. How does remote work affect salary and opening

```python
opening_count = df['remote_type'].value_counts()

print(opening_count)

opening_count.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Job Openings by Work Type')
plt.xlabel('Work Type')
plt.ylabel('Number of Openings')
plt.show()
```

```python
avg_salary = (
    df.groupby('remote_type')['salary']
      .mean()
      .sort_values(ascending=False)
)

print(avg_salary)

avg_salary.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Salary by Work Type')
plt.xlabel('Work Type')
plt.ylabel('Average Salary (LPA)')
plt.show()
```

```python
result = (
    df.groupby('remote_type')
      .agg(
          Avg_Salary=('salary', 'mean'),
          Openings=('remote_type', 'count')
      )
      .reset_index()
)

print(result)
```

# 7. how has job demand changed from 2020- 2026 ?

```python
df.head()
```

```python
job_demand_trend = df.groupby('job_posting_year')['job_openings'].sum().reset_index()

plt.figure(figsize=(10,6))

plt.plot(
    job_demand_trend['job_posting_year'],
    job_demand_trend['job_openings'],
    marker='o',
    linewidth=3
)

plt.title('Job Demand Trend (2020-2026)')
plt.xlabel('Year')
plt.ylabel('Total Number of Job Openings')
plt.grid(True)

plt.show()
```

```python
growth = (
    (df['job_posting_year'].iloc[-1] -
     df['job_openings'].iloc[0])
    /
    df['job_openings'].iloc[0]
) * 100

print(f"Growth from 2020 to 2026: {growth:.2f}%")
```
